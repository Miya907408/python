```python
#get the 3rd element and 3rd element from last of a tuple
tuple1=(10,20,30,40,50,60)
print("3rd element:",tuple1[2])
print("3rd element from last:",tuple1[-3])
```

```python
#Accept comma-seperated numbers,create a tuple and find average
numbers=input("Enter numbers seperated by comma:")
num_tuple=tuple(int(x) for x in numbers.split(","))
average=sum(num_tuple)/len(num_tuple)
print("Tuple:",num_tuple)
print("Average:",average)
```
