```python
#Create a list of colors and display first and last colors
colors=input("Enter colors seperated by comma:")
color_list=colors.split(",")
print("First color:",color_list[0])
print("Last color:",color_list[-1])
```

```python
#Accept filename and print extension
filename=input("Enter filename:")
extension=filename.split(".")[-1]
print("Extension:",extension)
```

```python
#Accept comma-seperated numbers,create a list and find the sum
numbers=input("Enter numbers seperated by comma:")
num_list=[int(x)for x in numbers.split(",")]
print("List:",num_list)
print("Sum:",sum(num_list))
```
