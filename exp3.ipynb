```python
#Exchange first and last characters
s=input("Enter a string:")
new=s[-1]+s[1:-1]+s[0]
print("New string:",new)
```

```python
#first 2 and last 2 characters
s=input("Enter a string:")
new=s[:2]+s[-2:]
print("New string:",new)
```

```python
#Swap character at position1 of two strings
s1=input("Enter first string:")
s2=input("Enter second string:")
new1=s2[0]+s1[1:]
new2=s1[0]+s2[1:]
print("After swap:",new1,new2)
```

```python
#change all occurences of first character to $,expect the first
s=input ("Enter a string:")
new=s[0]+s[1:].replace(s[0],'$')
print("New string:",new)
```
