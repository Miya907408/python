```python
#sum of number
num=int(input("Enter a number:"))
num1=num
num2=num*10+num1
num3=num*100+num2
print(num1,"+",num2,"+",num3,"=",num1+num2+num3)
```

```python
#Area of circle
from math import pi
r=int(input("Enter the radius of a circle:"))
area=pi*r*r
print("Area of circle:",f"{area:.2f}")
```
