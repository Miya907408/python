```python
#colorset1 not contained in colorset2
input1=input("Enter colors for set 1(seperated by commas):")
input2=input("Enter colors for set 2(seperated by commas):")
color_set1={color.strip()for color in input1.split(",")}
color_set2={color.strip()for color in input2.split(",")}
result=color_set1-color_set2
print("Colors in set 1 not in set 2:",result)
```

```python
#combine all colors from set1and set2
input1=input("Enter colors for set 1(seperated by commas):")
input2=input("Enter colors for set 2(seperated by commas):")
color_set1={color.strip()for color in input1.split(",")}
color_set2={color.strip()for color in input2.split(",")}
result=color_set1 | color_set2
print("All unique colors from both sets:",result)
```
